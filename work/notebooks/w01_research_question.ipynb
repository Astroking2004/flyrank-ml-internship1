## 3. Quick look at the data (2-3 real numbers)

I will show 2–3 real numbers from the starter dataset that support this lane and make the problem feel worth pursuing. These numbers should be concrete, easy to explain, and relevant to why the refresh-prioritization idea is promising.

```python
import pandas as pd
from pathlib import Path

path = Path('content_refresh_anonymized.csv')
df = pd.read_csv(path)

rows = df.shape[0]
cols = df.shape[1]
ctr_low_share = (df['ctr'] < 0.2).mean()
negative_trend_share = (df['trend_pct'] < 0).mean()
avg_position_zero_count = int((df['avg_position'] == 0).sum())
median_ctr = df['ctr'].median()
median_days_with_impressions = int(df['days_with_impressions'].median())

summary = {
    'rows': rows,
    'columns': cols,
    'share_ctr_below_0.2': round(ctr_low_share, 4),
    'share_negative_trend': round(negative_trend_share, 4),
    'avg_position_zero_count': avg_position_zero_count,
    'median_ctr': round(median_ctr, 4),
    'median_days_with_impressions': median_days_with_impressions,
}

print(summary)
print('top_content_types')
print(df['content_type'].value_counts().head().to_string())
```
